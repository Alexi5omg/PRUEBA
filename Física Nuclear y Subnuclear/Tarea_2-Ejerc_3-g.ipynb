```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Definimos las funciones t_MAX y t_MIN (expresiones calculadas en el inciso f)
def t_max(M,m1,m2,m3,s):
    A = (M**2 + s - m1**2)*(s + m2**2 - m3**2)/(2*s)
    B = ( ((M**2 + s - m1**2)**2 - 4*M**2*s)**(1/2) * ((s + m2**2 - m3**2)**2 - 4*s*m2**2)**(1/2) )/(2*s)
    return M**2 + m2**2 - A + B*(+1)

def t_min(M,m1,m2,m3,s):
    A = (M**2 + s - m1**2)*(s + m2**2 - m3**2)/(2*s)
    B = ( ((M**2 + s - m1**2)**2 - 4*M**2*s)**(1/2) * ((s + m2**2 - m3**2)**2 - 4*s*m2**2)**(1/2) )/(2*s)
    return M**2 + m2**2 - A + B*(-1)

# Definimos otras funciones para s_MAX y s_MIN (expresiones calculadas en el inciso b)
def s_min(m2,m3):
    return (m2 + m3)**2

def s_max(M,m1):
    return (M - m1)**2
```

```python
M_1 = 4
m1_1 = 1
m2_1 = m1_1
m3_1 = m1_1
s_1 = np.linspace(s_min(m2_1,m3_1),s_max(M_1,m1_1),150)

plt.plot(s_1,t_max(M_1,m1_1,m1_1,m1_1,s_1),'.-',c='teal',label=r't$_{\text{máx}}$')
plt.plot(s_1,t_min(M_1,m1_1,m1_1,m1_1,s_1),'.-',c='darkred',label=r't$_{\text{mín}}$')
plt.title(r'Límite del espacio fase s-t')
plt.xlabel(r's [$GeV^2$]'), plt.ylabel(r't [$GeV^2$]')
plt.legend()
plt.grid()
plt.show()
```

```python
M_2 = 4
m1_2 = 2.5
m2_2 = 0.5
m3_2 = m2_2
s_2 = np.linspace(s_min(m2_2,m3_2),s_max(M_2,m1_2),150)

plt.plot(s_2,t_max(M_2,m1_2,m2_2,m3_2,s_2),'.-',c='teal',label=r't$_{\text{máx}}$')
plt.plot(s_2,t_min(M_2,m1_2,m2_2,m3_2,s_2),'.-',c='darkred',label=r't$_{\text{mín}}$')
plt.title(r'Límite del espacio fase s-t')
plt.xlabel(r's [$GeV^2$]'), plt.ylabel(r't [$GeV^2$]')
plt.legend()
plt.grid()
plt.show()
```
